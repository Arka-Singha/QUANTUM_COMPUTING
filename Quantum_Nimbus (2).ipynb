```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def transform_function(x, y):
    """Implements F(x,y) = [x, y, x**2 + y**2]"""
    return np.array([x, y, x**2 + y**2])

# Example data A and B (replace with your actual data)
A = np.random.randn(100, 2)
B = np.random.randn(100, 2)

# Transform
A1 = np.array([transform_function(x, y) for x, y in zip(np.ravel(A[:,0]), np.ravel(A[:,1]))])
B1 = np.array([transform_function(x, y) for x, y in zip(np.ravel(B[:,0]), np.ravel(B[:,1]))])

# Plot in 3D
fig = plt.figure(figsize=(10, 5))
ax = fig.add_subplot(121, projection='3d')
ax.set_title("Data in 3D (separable with hyperplane)")
ax.scatter(A1[:,0], A1[:,1], A1[:,2], marker='o')
ax.scatter(B1[:,0], B1[:,1], B1[:,2], marker='s', c='C3')
ax.view_init(5, 68)

# Hyperplane in 3D
x = np.linspace(-1.25, 1.25, 10)
y = np.linspace(-1.25, 1.25, 10)
X, Y = np.meshgrid(x, y)
Z = 0.5 * np.ones_like(X)
ax.plot_surface(X, Y, Z, color='#343A3F', alpha=0.5)

# Project data to 2D
ax2d = fig.add_subplot(122)
ax2d.set_title("Data in 2D (with hyperplane projection)")
ax2d.scatter(A1[:,0], A1[:,1], marker='o')
ax2d.scatter(B1[:,0], B1[:,1], marker='s', c='C3')

ax2d.add_patch(plt.circle((0.0), radius=np.sqrt(0.5), fill = False, linestye = 'solid', linewidth = 4.8, color='343A3F'))

plt.show()
```

```python
def grover_oracle(marked_states):
    """Build a Grover oracle for multiple marked states

    Here we assume all input marked states have the same number of bits

    Parameters:
        marked_states (str or list): Marked states of oracle

    Returns:
        QuantumCircuit: Quantum circuit representing Grover oracle
    """
    if not isinstance(marked_states, list):
        marked_states = [marked_states]
    # Compute the number of qubits in circuit
    num_qubits = len(marked_states[0])

    qc = QuantumCircuit(num_qubits)
    # Mark each target state in the input list
    for target in marked_states:
        # Flip target bit-string to match Qiskit bit-ordering
        rev_target = target[::-1]
        # Find the indices of all the '0' elements in bit-string
        zero_inds = [ind for ind in range(num_qubits) if rev_target.startswith("0", ind)]
        # Add a multi-controlled Z-gate with pre- and post-applied X-gates (open-controls)
        # where the target bit-string has a '0' entry
        qc.x(zero_inds)
        qc.compose(MCMT(ZGate(), num_qubits - 1, 1), inplace=True)
        qc.x(zero_inds)
    return qc
```

```python
marked_states = ["011","100"]

oracle = grover_oracle(marked_states)
oracle.draw(output="mpl")
```

```python
grover_op = GroverOperator(oracle)
grover_op.decompose().draw(output = "mpl")
```

```python
def c_amod15(a):
    """
    Controlled Multiplication by a mod 15.
    This is hard coded for simplicity.
    """
    if a not in [2, 4, 7, 8, 11, 13]:
        raise valueError("a must not have common factors with 15")
    U = QuantumCircuit(4)
    if a in [2, 13]:
        U.swap(2, 3)
        U.swap(1, 2)
        U.swap(0, 1)
    if a in [7, 8]:
        U.swap(0, 1)
        U.swap(1, 2)
        U.swap(2, 3)
    if a in [4, 11]:
        U.swap(1, 3)
        U.swap(0, 2)
    if a in [7, 11, 13]:
        for q in range(4):
            U.x(q)
    U = U. to_gate()
    U.name = f"{a} mod 15"
    c_U = U.control()
    return c_U
```

```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit.circuit.library import QFT
from qiskit.primitives import Sampler

def phase_estimator(
    controlled_operation: QuantumCircuit,
    psi_prep: QuantumCircuit,
    precision: int
):
    """ 
    Carry out phase estimation on a simulator.
    Args:
        controlled_operation (QuantumCircuit): The operation to perform phase estimation on,
                                               controlled by one qubit.
        psi_prep (QuantumCircuit): Circuit to prepare |psi>
        precision (int): Number of counting qubits to use.
    Returns:
        float: Best guess for phase of U |psi>
    """
    control_register = QuantumRegister(precision)
    output_register = ClassicalRegister(precision)
    target_register = QuantumRegister(psi_prep.num_qubits)
    
    qc = QuantumCircuit(control_register, target_register, output_register)

    # Prepare |psi>
    qc.compose(psi_prep, qubits=target_register, inplace=True)

    # Do phase estimation
    for index, qubit in enumerate(control_register):
        qc.h(qubit)  # Apply Hadamard gates to control qubits
        for _ in range(2**index):
            qc.compose(
                controlled_operation,
                qubits=[qubit] + list(target_register),
                inplace=True,
            )

    # Apply the inverse Quantum Fourier Transform (QFT)
    qc.compose(QFT(precision, inverse=True), qubits=control_register, inplace=True)

    # Measure the control register
    qc.measure(control_register, output_register)

    # Run the circuit on a simulator and retrieve the measurement result
    measurement = Sampler().run(qc, shots=1).result().quasi_dists[0].popitem()[0]

    return measurement / 2**precision
```

```python
# AND FINALLY WE CAN RUN THE CIRCUIT TO TRY TO FIND A NATIONAL FACTOR OF 15.

from fractions import Fraction
from math import gcd

a = 8
N = 15

FACTOR_FOUND = False
ATTEMPT = 0
while not FACTOR_FOUND:
    ATTEMPT += 1
    print(f"\nAttempt {ATTEMPT}")

    phase = phase_estimator(
        c_amod15(a),
        psi_prep,
        precision=8
    )
    frac = Fraction(phase).limit_denominator(N)
    r = frac.denominator
    if phase != 0:
        # Guess for a factor is gcd(x^{r/2} -1, 15)
        guess = gcd(a**(r//2) - 1, N)
        if guess not in [1, N] and (N % guess) == 0:
            # Guess is a factor !
            print(f"Non-trivial factor found: {guess}")
            FACTOR_FOUND = True
```

```python
# Quantum Error Corrections

from qiskit import QuantumCircuit, ClassicalRegister, QuantumRegister, transpile
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram

# Creating registers
qr = QuantumRegister(5, name='q')  # Use QuantumRegister for qubits
cr0 = ClassicalRegister(1, name='c0')  # Use ClassicalRegister for classical bits
cr1 = ClassicalRegister(1, name='c1')
cr2 = ClassicalRegister(1, name='c2')

# Create the quantum circuit
bit_flip = QuantumCircuit(qr, cr0, cr1, cr2)  # Specify registers directly in the circuit

# Prepares qubit in the desired initial state 
bit_flip.h(qr[0])

# Encodes the qubit in a three-qubit entangled state  
bit_flip.cx(qr[0], qr[1])
bit_flip.cx(qr[0], qr[2])

# Bit-flip error on the second qubit
bit_flip.x(qr[1])

# Adds additional two qubits for error-correction
bit_flip.cx(qr[0], qr[3])
bit_flip.cx(qr[1], qr[3])
bit_flip.cx(qr[0], qr[4])
bit_flip.cx(qr[2], qr[4])

# Measure the two additional qubits
bit_flip.measure(qr[3], cr0[0])
bit_flip.measure(qr[4], cr1[0])

# Do error correction
bit_flip.x(qr[1]).c_if(cr0, 1)
bit_flip.x(qr[2]).c_if(cr1, 1)

# Decodes the qubit from the three-qubit entangled state  
bit_flip.cx(qr[0], qr[1])
bit_flip.cx(qr[0], qr[2])

# Check the state of the initial qubit
bit_flip.measure(qr[0], cr2[0])

# Shows gates of the circuit
circuits = ['bit_flip']
# Get the Aer simulator backend
simulator = Aer.get_backend('qasm_simulator')
transpiled_circuit = transpile(bit_flip, backend=simulator)
job = simulator.run(transpiled_circuit, shots=1024)

# Get the results and plot the histogram
result = job.result()
counts = result.get_counts(transpiled_circuit)
plot_histogram(counts)
bit_flip.draw()
```
